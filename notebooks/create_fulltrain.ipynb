```python
import pandas as pd
from pathlib import Path

POS_PATH = Path("/media/auk/projects/gak76/vira_beg_outputs/training_data/positives/positive_vira_1s_gk.pkl")
XC_PATH = Path("/media/auk/projects/gak76/vira_beg_outputs/training_data/negative_multihot_labels_gk.pkl")
FIELD_PATH = Path("/media/auk/projects/gak76/vira_beg_outputs/training_data/field_negatives/field_negatives_gk.pkl")

pos_df = pd.read_pickle(POS_PATH)
xc_df = pd.read_pickle(XC_PATH)
field_df = pd.read_pickle(FIELD_PATH)

print("POSITIVES")
print(type(pos_df))
print(pos_df.shape)
print(pos_df.columns if hasattr(pos_df, "columns") else "no columns")
display(pos_df.head())

print("\nXC")
print(type(xc_df))
print(xc_df.shape)
print(xc_df.columns if hasattr(xc_df, "columns") else "no columns")
display(xc_df.head())

print("\nFIELD")
print(type(field_df))
print(field_df.shape)
print(field_df.columns if hasattr(field_df, "columns") else "no columns")
display(field_df.head())
```

```python
MERGE_DIR = Path("/media/auk/projects/gak76/vira_beg_outputs/training_data/fulltrain")
MERGE_DIR.mkdir(parents=True, exist_ok=True)

pos_df = pd.read_pickle(POS_PATH)
xc_df = pd.read_pickle(XC_PATH)
field_df = pd.read_pickle(FIELD_PATH)

print("Loaded:")
print("positives:", pos_df.shape)
print("xc:", xc_df.shape)
print("field:", field_df.shape)

# ----------------------------
# normalize positives
# ----------------------------
if isinstance(pos_df.index, pd.MultiIndex):
    pos_df = pos_df.reset_index()

pos_df["virail"] = 1

# ----------------------------
# normalize XC negatives
# ----------------------------
if isinstance(xc_df.index, pd.MultiIndex):
    xc_df = xc_df.reset_index()

xc_df["virail"] = 0

if "end_time" not in xc_df.columns and "start_time" in xc_df.columns:
    xc_df["end_time"] = xc_df["start_time"] + 1.0

# ----------------------------
# merge positives + XC
# ----------------------------
main_cols = sorted(set(pos_df.columns) | set(xc_df.columns))
for col in main_cols:
    if col not in pos_df.columns:
        pos_df[col] = 0
    if col not in xc_df.columns:
        xc_df[col] = 0

pos_df = pos_df[main_cols]
xc_df = xc_df[main_cols]

main_df = pd.concat([pos_df, xc_df], ignore_index=True).fillna(0)

# ----------------------------
# normalize field negatives
# ----------------------------
if isinstance(field_df.index, pd.MultiIndex):
    field_df = field_df.reset_index()

if "full_path" in field_df.columns and "file" not in field_df.columns:
    field_df = field_df.rename(columns={"full_path": "file"})

field_df["virail"] = 0
field_df["field_data"] = 1

if "end_time" not in field_df.columns and "start_time" in field_df.columns:
    field_df["end_time"] = field_df["start_time"] + 1.0

# ----------------------------
# align columns and append field negatives
# ----------------------------
all_cols = sorted(set(main_df.columns) | set(field_df.columns))
for col in all_cols:
    if col not in main_df.columns:
        main_df[col] = 0
    if col not in field_df.columns:
        field_df[col] = 0

main_df = main_df[all_cols]
field_df = field_df[all_cols]

full_df = pd.concat([main_df, field_df], ignore_index=True).fillna(0)

# ----------------------------
# final OpenSoundscape format
# ----------------------------
full_df = full_df.set_index(["file", "start_time", "end_time"]).sort_index()

print("Final full dataset shape:", full_df.shape)
display(full_df.head())

full_df.to_pickle(MERGE_DIR / "full_dataset_1s_gk.pkl")
print("Saved to:", MERGE_DIR / "full_dataset_1s_gk.pkl")
```
